You can fix it by manipulating the shape of one of our tensors, using **transpose**

tensor_A.T

A transpose switches the axes or dimensions of a given tensor

```python
tensor_A.T @ tensor_B
```

```python
tensor_A @ tensor_B.T #These are different, see?
```

```python
tensor_B.T @ tensor_A #Even this one is still a bit different, even if it's the same matrix, transposed.
```

```python
tensor_BAT = tensor_B @ tensor_A.T
print(tensor_BAT)
print(tensor_BAT.T)
print(tensor_A @ tensor_B.T)
```

"In computer programming, the stride of an array (also referred to as increment, pitch or step size) is the number of locations in memory between beginnings of successive array elements, measured in bytes or in units of the size of the array's elements. The stride cannot be smaller than the element size but can be larger, indicating extra space between elements.

An array with stride of exactly the same size as the size of each of its elements is contiguous in memory. Such arrays are sometimes said to have unit stride. Unit stride arrays are sometimes more efficient than non-unit stride arrays, but non-unit stride arrays can be more efficient for 2D or multi-dimensional arrays, depending on the effects of caching and the access patterns used. ### What is a Tensor?

## Vectors
You may think of it as an arrow with magnitude and direction.
orientation - direction
magnitude - length of arrow

It may represent velocity, strength, ...
Vectors can represent an AREA as well, but how?

You simply make the length of the vector proportional to the area, and then you make the direction of the arrow to be PERPENDICULAR to the square.

Vectors are member of a wider class of object named TENSORS.

## Vector Components and Basis Vectors

Look at the three sticks that usually are taken to represent the X,Y,Z axes. It represents a COORDINATE SYSTEM. The Cartesian Coordinate System.

Coordinate systems come with basis vectors, or UNIT VECTORS. They have a length of 1

1 is whatever the unit you're expressing your vectors in.
x_hat or i_hat. It points in the direction of increasing X coordinates

y_hat
z_hat or k_hat

you get the idea. Coordinate System; Unit Vectors.

Now you can find the COMPONENTS of your vectors.

## Vector Components
Finding Vector Components.

Begin with a vector in the XY plane. To find the X component, project the vector into the X axis. To find the Y component, project the vector into the Y axis.

You 'shine a light' perpendicular to the X and Y axes, respectively. Notice that the inclination of the arrow changes the length of the shadow.

Ask yourself, to get from the base to the tip of the vector, how far do I have to go in the X vector, and how far in the Y vector.
How many x_hat units, how many y_hat units? (which are the basis vectors)

That means that instead of drawing an arrow for this vectors, you can say
4 x_hat, and 3 y_hat.

This is a perfectly valid representation of the vectors.
A = [[4], [3], [0]]

## Combinations of Components and Basis Vectors
All observers in all reference frames agree, not on basis vectors or components, but on their combination.

That is because basis vectors transform one way between reference frames.

AND the components transform in just such a way so as to keep the combination of these the same for all observers.

These characteristics led Lillian Lieber to call tensors the facts of the universe.

```python
batch_size = 64

#Create data loaders
train_dataloader = DataLoader(
    dataset=training_data,
    batch_size=batch_size
)

test_dataloader = DataLoader(
    dataset=test_data,
    batch_size=batch_size
)

for X, y in test_dataloader:
  print(f"Shape of X [N, C, H, W]: {X.shape}")
  print(f"Shape of y: {y.shape}, {y.dtype}")
  break
```

### Creating a model
To define a Neural Network in PyTorch, we create a class that inherits from nn.Module

We define the layers of the network in the '__init__' function,
and specify how data will pass through the network in the 'forward' function.

To accelerate operations in the NN, we move it to the Accelerator, such as CUDA, MPS, MTIA, or XPU.

Not the device-agnostic code, which will default to CPU if no Accelerator is available.

```python
device = torch.accelerator.current_accelerator().type if torch.accelerator.is_available() else "cpu"
print(f"Using {device} device")
```

```python
class NeuralNetwork(nn.Module):
  def __init__(self):
    super().__init__()
    self.flatten = nn.Flatten()
    self.linear_relu_stack = nn.Sequential( #This must be like the API for creating an acyclic graph
      nn.Linear(in_features=28*28, out_features=512), #Images are 28*28. And the output is a design choice.
      nn.ReLU(),
      nn.Linear(512, 512),
      nn.ReLU(),
      nn.Linear(512,10)
    )

  def forward(self, x):
    x = self.flatten(x)
    logits = self.linear_relu_stack(x)
    return logits

model = NeuralNetwork().to(device)
print(model)
```

### Optimizing Model Parameters
To train a model, we need:
- a Loss Function
- an Optimizer

```python
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=1e-3
)
```

In a Single Training Loop, the model makes predictions on the Training Dataset (fed to it in batches), and backpropagates the prediction error to adjust the model's parameters.

```python
def train(dataloader, model, loss_fn, optimizer):
  size = len(dataloader.dataset)
  model.train()
  for batch, (X, y) in enumerate(dataloader):
    X, y = X.to(device), y.to(device)

    #Compute prediction error
    pred = model(X)
    loss = loss_fn(pred, y)

    # Backpropagation
    loss.backward() #Populates model's .grad fields, preparing for the optimization!
    optimizer.step()  # Uses the gradients (.grad field, calculated by the loss function) to update the parameters
    optimizer.zero_grad() # We reset for the next iteration (reset .grad)

    if batch % 100 == 0:
      loss, current = loss.item(), (batch + 1) * len(X) #Current loss, and number of samples processed so far (batch# * length of current batch, all are the same length)
      print(f"loss: {loss:>7f} [{current:>5d}/{size:>5d}]")
```

We must also check the model's performance against the Test Dataset to ensure it is learning

```python
def test(dataloader, model, loss_fn):
  size = len(dataloader.dataset)
  num_batches = len(dataloader) #In PyTorch it's common to have two DataLoaders: one for the Training Dataset, one for the Test Dataset.
  model.eval() #Tells your model to change from TRAINING MODE to EVALUATION MODE. The behavior of some layers changes (Dropout, BatchNorm). We don't seem to have any of these, but it might have been included as an example!
  test_loss, correct = 0, 0
  with torch.no_grad(): #Disable automatic gradient tracking. PyTorch records the sequence of operations (like addition, multiplication, layers, etc.) as a graph. Then loss.backward() is a backward traversal of this graph.
    for X, y in dataloader:
      X, y = X.to(device), y.to(device)
      pred = model(X)
      test_loss += loss_fn(pred, y).item()
      correct += (pred.argmax(1) == y).type(torch.float).sum().item()

  test_loss /= num_batches # Loss per batch I think?
  correct /= size #Would give you a ratio of correct guesses
  print(f"Test Error: \n Accuracy: {(100*correct):>0.1f}%, Avg Loss: {test_loss:>8f} \n")
```

The training process is conducted over several iterations (epochs). During each epoch, the model learns PARAMETERS to make better predictions.

we print the model's accuracy and loss at each epoch; we'd like to see the acuracy INCREASE, and the loss DECREASE with every epoch

```python
epochs = 5
for t in range(epochs):
  print(f"Epoch {t+1}\n-----------------------")
  train(
      dataloader=train_dataloader,
      model=model,
      loss_fn=loss_fn,
      optimizer=optimizer
  )
  test(
      test_dataloader,
      model,
      loss_fn
  )
print("Done!")
```
